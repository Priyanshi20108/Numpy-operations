### numpy 
NumPy is a core Python library for numerical computing, built for handling large arrays and matrices efficiently. It is significantly faster than Python's built-in lists because it uses optimized C language style storage where actual values are stored at contiguous locations

```python
## importing library
import numpy as np
```

```python
print(np.__version__)   ## checking numpy version
```

```python
## creating nd array
arr = np.array([1,2,3,4,5])
print(arr)
```

```python
# 2d array
l1 = [1,2,3,4]
l2 = [5,6,7,8]
arr2 = np.array([l1,l2])
print(arr2)
```

```python
#3d array
l1 = [1,2,3,4]
l2 = [5,6,7,8]
l3 = [9,10,22,34]

arr3 = np.array([[l1,l2,l3]])
print(arr3)
```

```python
np.ones((2,3))          ## creates an array of ones
```

```python
np.zeros((2,3))     ## creates an array of zeros
```

```python
np.linspace(0,1,5)          ## creates evenly spaced values
```

```python
np.eye(3)                ## 3*3 Identiy matrix
```

```python
np.arange(10)          ## creates range of value
```

### Indexing and Slicing

```python
arr[1:3]             ## access elements
```

```python
## in 2d 
arr2[:, 1:3]
```

```python
## in 3d
arr3[:,:,1:3]
```

```python
## boolean indexing 
arr[arr > 2]     ## filter with condition
```

```python
data = np.array([15 , 22 , 9 , 34 , 18 , 5])
```

```python
## combining conditions (& for AND , | for OR)
data[(data >= 10) & (data <= 20)]
```

```python
## conditional replacement
np.where(data > 15 , "High" , "Low")
```

### Mathematical Operations

```python
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
```

```python
np.add(a,b)           ## addition
          ## or 
#a + b
```

```python
np.subtract(b,a)     ## subtraction
         ## or 
# b - a
```

```python
np.multiply(a,b)       ## multiplication
    ## or 
# a * b
```

```python
np.divide(a,b)      ## division
     ## or
# a / b
```

```python
np.power(a,2)        ## exponention
      ## or 
# a ** 2
```

```python
np.sqrt(a)        ## square root
```

```python
np.dot(l1 , l2)           ## matrix multiplication
```

### Handling missing data

```python
np.isnan(nan_arr)     ## checks nan values
```

```python
## saf aggregations ingoring NaNs
```

```python
np.nansum(nan_arr)    
```

```python
np.nanmean(nan_arr)
```

### Random number generation

```python
np.random.seed(42)          ## set seed for reproducibility 
```

```python
np.random.rand(3,3)         ## uniform distribution
```

```python
np.random.randn(10)     ## standard normal distribution
```

### Set Operations & Unique Values 

```python
words = np.array(["apple" , "banana" , "apple" , "banana" , "orange"])
```

```python
np.unique(words)      ## unique elements
```

```python
np.unique(words , return_counts = True )          ## unique elements + frequencies
```

```python
s1 = np.array([1,2,3,4,5])
s2 = np.array([2,4,5,6,9])
```

```python
np.intersect1d(s1 , s2)     ## common elements
```

```python
np.setdiff1d(s1 , s2)      ## in s1 but not in s2
```

```python
np.setdiff1d(s2 , s1)      ## in s2 but not in s1
```
