### numpy 
NumPy is a core Python library for numerical computing, built for handling large arrays and matrices efficiently. It is significantly faster than Python's built-in lists because it uses optimized C language style storage where actual values are stored at contiguous locations

```python
## importing library
import numpy as np
```

```python
print(np.__version__)   ## checking numpy version
```

```python
## creating nd array
arr = np.array([1,2,3,4,5])
print(arr)
```

```python
# 2d array
l1 = [1,2,3,4]
l2 = [5,6,7,8]
arr2 = np.array([l1,l2])
print(arr2)
```

```python
#3d array
l1 = [1,2,3,4]
l2 = [5,6,7,8]
l3 = [9,10,22,34]

arr3 = np.array([[l1,l2,l3]])
print(arr3)
```

```python
print(arr2.shape , arr3.shape)           ## returns shape (rows,columns)
```

```python
print([arr2.ndim] , [arr3.ndim])        ## number of dimensions
```

```python
arr3.size    ## total elements
```

```python
arr3.dtype         ## datatype of an array
```

```python
np.zeros((2,3))     ## creates an array of zeros
```

```python
np.ones((2,3))          ## creates an array of ones
```

```python
np.arange(10)          ## creates range of value
```

```python
np.linspace(0,1,5)          ## creates evenly spaced values
```

## arrray reshaping 

```python
arr.reshape(5,1)          ## change array shape
```

```python
arr2.flatten()      ## converts to 1d
```

```python
arr2.ravel()        ## returns flatten array (changes original value)
```

## indexing and slicing

```python
arr[1:3]             ## access elements
```

```python
## in 2d 
arr2[:, 1:3]
```

```python
## in 3d
arr3[:,:,1:3]
```

```python
## boolean indexing 
arr[arr > 2]     ## filter with condition
```

## Mathematical Operations

```python
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])
```

```python
np.add(a,b)           ## addition
```

```python
np.subtract(b,a)     ## subtraction
```

```python
np.multiply(a,b)       ## multiplication
```

```python
np.divide(a,b)      ## division
```

```python
np.power(a,2)        ## exponention
```

```python
np.sqrt(a)        ## square root
```

### Statistical Operations

```python
a = np.array([1,2,3,4])
```

```python
print(np.mean(a))      ## mean
print(np.median(a))    ## median
print(np.sum(a))       ## sum of elements
print(np.min(a))       ## minimum value
print(np.max(a))       ## maximum value
```
